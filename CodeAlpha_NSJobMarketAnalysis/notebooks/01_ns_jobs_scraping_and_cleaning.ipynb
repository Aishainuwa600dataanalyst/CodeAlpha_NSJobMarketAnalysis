```python
def scrape_job(job_url):

    response = requests.get(job_url)

    job_soup = BeautifulSoup(response.text, "html.parser")

    job_title = job_soup.find(id="heading").get_text(" ", strip=True)

    employer = job_soup.find(id="employer_name").get_text(" ", strip=True)

    closing_date = job_soup.find(id="closing_date").get_text(" ", strip=True)
    closing_date = closing_date.replace("The closing date is ", "")

    date_posted = job_soup.find(id="date_posted").get_text(" ", strip=True)

    salary = job_soup.find(id="range_salary").get_text(" ", strip=True)

    contract_type = job_soup.find(id="contract_type").get_text(" ", strip=True)

    working_pattern = job_soup.find(id="working_pattern_heading").find_next_sibling("p").get_text(" ", strip=True)

    location = job_soup.find(id="employer_town").get_text(" ", strip=True)

    postcode = job_soup.find(id="employer_postcode").get_text(" ", strip=True)

    return {
        "job_title": job_title,
        "employer": employer,
        "location": location,
        "postcode": postcode,
        "salary": salary,
        "date_posted": date_posted,
        "closing_date": closing_date,
        "contract_type": contract_type,
        "working_pattern": working_pattern,
        "job_url": job_url
    }
```

```python
def get_text(soup, element_id):
    element = soup.find(id=element_id)

    if element:
        return element.get_text(" ", strip=True)

    return "Not available"


def scrape_job(job_url):

    response = requests.get(job_url)

    job_soup = BeautifulSoup(response.text, "html.parser")

    job_title = get_text(job_soup, "heading")

    employer = get_text(job_soup, "employer_name")

    closing_date = get_text(job_soup, "closing_date")
    closing_date = closing_date.replace("The closing date is ", "")

    date_posted = get_text(job_soup, "date_posted")

    salary = get_text(job_soup, "range_salary")

    contract_type = get_text(job_soup, "contract_type")

    working_heading = job_soup.find(id="working_pattern_heading")

    if working_heading:
        working_pattern = working_heading.find_next_sibling("p").get_text(
            " ", strip=True
        )
    else:
        working_pattern = "Not available"

    location = get_text(job_soup, "employer_town")

    postcode = get_text(job_soup, "employer_postcode")

    return {
        "job_title": job_title,
        "employer": employer,
        "location": location,
        "postcode": postcode,
        "salary": salary,
        "date_posted": date_posted,
        "closing_date": closing_date,
        "contract_type": contract_type,
        "working_pattern": working_pattern,
        "job_url": job_url
    }
```

```python
# Clean text fields

df["salary"] = df["salary"].str.replace(r"\s+", " ", regex=True).str.strip()

df["working_pattern"] = df["working_pattern"].str.replace(r"\s+", " ", regex=True).str.strip()

df["job_title"] = df["job_title"].str.replace(r"\s+", " ", regex=True).str.strip()

df["employer"] = df["employer"].str.replace(r"\s+", " ", regex=True).str.strip()

df["location"] = df["location"].str.replace(r"\s+", " ", regex=True).str.strip()

print("Cleaning complete!")
```

```python
import re

df["salary_min"] = df["salary"].apply(
    lambda x: int(re.search(r"£([\d,]+)", x).group(1).replace(",", ""))
    if re.search(r"£([\d,]+)", x)
    else None
)

df["salary_max"] = df["salary"].apply(
    lambda x: int(re.findall(r"£([\d,]+)", x)[-1].replace(",", ""))
    if len(re.findall(r"£([\d,]+)", x)) >= 2
    else None
)

print(df[["job_title", "salary", "salary_min", "salary_max"]])
```

```python
# Convert posting date to a proper date
df["date_posted"] = pd.to_datetime(df["date_posted"], errors="coerce")

# Convert closing date to a proper date where possible
df["closing_date_clean"] = pd.to_datetime(
    df["closing_date"],
    errors="coerce"
)

# Calculate the midpoint of each salary range
df["salary_midpoint"] = (
    df["salary_min"] + df["salary_max"]
) / 2

# Show the important analysis columns
print(df[
    [
        "job_title",
        "employer",
        "location",
        "salary_min",
        "salary_max",
        "salary_midpoint",
        "contract_type",
        "date_posted"
    ]
])
```

```python
from urllib.parse import urljoin
import time

base_url = "https://www.jobs.nhs.uk"

all_job_urls = []

for page in range(1, 11):
    print(f"Collecting page {page}/10...")

    page_url = (
        "https://www.jobs.nhs.uk/candidate/search/results"
        f"?searchFormType=main&keyword=Data%20Analyst&language=en&page={page}"
    )

    response = requests.get(page_url)
    page_soup = BeautifulSoup(response.text, "html.parser")

    links = page_soup.find_all(
        "a",
        href=lambda x: x and "/candidate/jobadvert/" in x
    )

    for link in links:
        full_url = urljoin(base_url, link["href"])

        if full_url not in all_job_urls:
            all_job_urls.append(full_url)

    time.sleep(1)

print("\nTotal unique job URLs collected:", len(all_job_urls))
```

```python
jobs_data_100 = []

for i, url in enumerate(all_job_urls, start=1):
    print(f"Scraping job {i}/100...", end="\r")

    try:
        job = scrape_job(url)
        jobs_data_100.append(job)

    except Exception as e:
        print(f"\nCould not scrape job {i}: {e}")

    time.sleep(1)

print("\n\nTotal jobs successfully scraped:", len(jobs_data_100))
```

```python
df = pd.DataFrame(jobs_data_100)

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# Clean text fields in the 100-job dataset

text_columns = [
    "job_title",
    "employer",
    "location",
    "postcode",
    "salary",
    "closing_date",
    "contract_type",
    "working_pattern"
]

for column in text_columns:
    df[column] = (
        df[column]
        .fillna("Not available")
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

print("Text cleaning complete!")
```

```python
import re

def extract_salary_min(salary):
    matches = re.findall(r"£([\d,]+(?:\.\d+)?)", salary)

    if matches:
        return float(matches[0].replace(",", ""))

    return None


def extract_salary_max(salary):
    matches = re.findall(r"£([\d,]+(?:\.\d+)?)", salary)

    if len(matches) >= 2:
        return float(matches[-1].replace(",", ""))

    return None


df["salary_min"] = df["salary"].apply(extract_salary_min)
df["salary_max"] = df["salary"].apply(extract_salary_max)

df["salary_midpoint"] = (
    df["salary_min"] + df["salary_max"]
) / 2

print(
    df[
        [
            "job_title",
            "salary",
            "salary_min",
            "salary_max",
            "salary_midpoint"
        ]
    ].head(10)
)
```

```python
# Check salary data quality

total_jobs = len(df)

salary_available = df["salary_min"].notna().sum()
salary_missing = df["salary_min"].isna().sum()

print("Total jobs:", total_jobs)
print("Jobs with salary data:", salary_available)
print("Jobs without salary data:", salary_missing)
```

```python
def classify_role(title):
    title = title.lower()

    if any(word in title for word in [
        "data analyst",
        "data engineer",
        "data scientist",
        "data lead",
        "data manager",
        "data quality",
        "data administrator",
        "data coordinator",
        "data assistant",
        "data officer",
        "data entry",
        "data development",
        "data protection",
        "data security",
        "information analyst",
        "information manager",
        "information assistant",
        "analytical"
    ]):
        return "Data & Analytics"

    elif any(word in title for word in [
        "cyber security",
        "cybersecurity",
        "information security"
    ]):
        return "Cybersecurity"

    elif any(word in title for word in [
        "it ",
        "ict",
        "application support",
        "network",
        "technology",
        "digital",
        "systems administrator",
        "service desk"
    ]):
        return "IT & Technology"

    elif any(word in title for word in [
        "finance",
        "financial",
        "commercial"
    ]):
        return "Finance & Commercial"

    elif any(word in title for word in [
        "research",
        "clinical audit",
        "bioinformatic",
        "clinical coding"
    ]):
        return "Clinical & Research"

    elif any(word in title for word in [
        "administrator",
        "administration",
        "secretary",
        "receptionist",
        "support officer"
    ]):
        return "Administration & Support"

    else:
        return "Other"


df["role_category"] = df["job_title"].apply(classify_role)

print(df["role_category"].value_counts())
```

```python
import re

# FIX (Oct 2026): "IT" is now matched as a whole word, so it no longer
# catches words that end in "it", like "Audit".
def classify_role(title):
    title = title.lower()

    # Cybersecurity first
    if any(word in title for word in [
        "cyber security",
        "cybersecurity",
        "information security"
    ]):
        return "Cybersecurity"

    # Finance and commercial
    elif any(word in title for word in [
        "finance",
        "financial",
        "commercial"
    ]):
        return "Finance & Commercial"

    # Business, performance and management analytics
    elif any(word in title for word in [
        "business analyst",
        "business and performance",
        "performance analyst",
        "performance manager",
        "portfolio analyst",
        "commissioning",
        "monitoring, evaluation",
        "merl",
        "service improvement",
        "operations lead"
    ]):
        return "Business & Performance"

    # Data and information roles
    elif any(word in title for word in [
        "data analyst",
        "data engineer",
        "data scientist",
        "data lead",
        "data manager",
        "data quality",
        "data administrator",
        "data coordinator",
        "data co-ordinator",
        "data assistant",
        "data officer",
        "data entry",
        "data development",
        "data protection",
        "data security",
        "information analyst",
        "information manager",
        "information assistant",
        "ehr",
        "him",
        "data input",
        "data care",
        "analyst - him",
        "analytical",
        "research data"
    ]):
        return "Data & Analytics"

    # IT and technology
    elif re.search(r"\bit\b", title) or any(word in title for word in [
        "ict",
        "im&t",
        "application support",
        "applications support",
        "network",
        "technology",
        "digital",
        "systems administrator",
        "service desk",
        "application analyst"
    ]):
        return "IT & Technology"

    # Clinical and research
    elif any(word in title for word in [
        "research",
        "clinical audit",
        "bioinformatic",
        "clinical coding",
        "clinical psychologist",
        "radiology"
    ]):
        return "Clinical & Research"

    # Administration and support
    elif any(word in title for word in [
        "administrator",
        "administration",
        "secretary",
        "receptionist",
        "support officer",
        "call handler",
        "coordinator"
    ]):
        return "Administration & Support"

    else:
        return "Other"


df["role_category"] = df["job_title"].apply(classify_role)

print(df["role_category"].value_counts())
```
